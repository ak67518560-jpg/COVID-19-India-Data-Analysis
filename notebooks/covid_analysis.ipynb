```python
#Missing Values Percentage
def missing_percentage(df):
    missing = df.isnull().sum()
    percentage = (missing / len(df)) * 100

    result = pd.DataFrame({
        "Missing Values": missing,
        "Missing Percentage": percentage
    })

    return result[result["Missing Values"] > 0].sort_values(
        by="Missing Percentage", ascending=False
    )

print("COVID Dataset")
display(missing_percentage(covid))

print("Vaccine Dataset")
display(missing_percentage(vaccine))

print("Testing Dataset")
display(missing_percentage(testing))
```

```python
# Date Conversion

covid["Date"] = pd.to_datetime(
    covid["Date"], dayfirst=True, errors="coerce"
)

vaccine["Updated On"] = pd.to_datetime(
    vaccine["Updated On"], dayfirst=True, errors="coerce"
)

testing["Date"] = pd.to_datetime(
    testing["Date"], dayfirst=True, errors="coerce"
)

print("Date Conversion Completed!")
```

```python
#Step 2: Check Date Range
print("COVID Date Range:")
print(covid["Date"].min(), "to", covid["Date"].max())

print("\nVaccine Date Range:")
print(vaccine["Updated On"].min(), "to", vaccine["Updated On"].max())

print("\nTesting Date Range:")
print(testing["Date"].min(), "to", testing["Date"].max())
```

```python
# Convert COVID columns into numeric

covid["ConfirmedIndianNational"] = pd.to_numeric(
    covid["ConfirmedIndianNational"],
    errors="coerce"
)

covid["ConfirmedForeignNational"] = pd.to_numeric(
    covid["ConfirmedForeignNational"],
    errors="coerce"
)

print("COVID Numeric Conversion Completed!")
```

```python
# Testing Numeric Conversion

testing["Negative"] = pd.to_numeric(
    testing["Negative"],
    errors="coerce"
)

testing["Positive"] = pd.to_numeric(
    testing["Positive"],
    errors="coerce"
)

testing["TotalSamples"] = pd.to_numeric(
    testing["TotalSamples"],
    errors="coerce"
)

print("Testing Numeric Conversion Completed!")
```

```python
## COVID Dataset Statistical Summary


display(covid[[
    "Confirmed",
    "Cured",
    "Deaths",
    "ConfirmedIndianNational",
    "ConfirmedForeignNational"
]].describe())
```

```python
## Total COVID Statistics

total_confirmed = covid["Confirmed"].max()
total_recovered = covid["Cured"].max()
total_deaths = covid["Deaths"].max()

print("Maximum Confirmed Cases:", total_confirmed)
print("Maximum Recovered Cases:", total_recovered)
print("Maximum Deaths:", total_deaths)
```

```python
# Recovery Rate and Death Rate

recovery_rate = (total_recovered / total_confirmed) * 100
death_rate = (total_deaths / total_confirmed) * 100
print("Recovery Rate:", round(recovery_rate, 2), "%")
print("Death Rate:", round(death_rate, 2), "%")
```

```python
# Check Unique State Names

print("Total Unique States:", covid["State/UnionTerritory"].nunique())
print("\nAll State Names:")
for state in sorted(covid["State/UnionTerritory"].dropna().unique()):
    print(state)
```

```python
# Display All Unique State Names

state_names = sorted(
    covid["State/UnionTerritory"].dropna().unique()
)

for i, state in enumerate(state_names, start=1):
    print(i, state)
```

```python
# Create Clean State Column

covid["State_Clean"] = covid["State/UnionTerritory"].str.strip()

# Correct spelling variations
state_mapping = {
    "Karanataka": "Karnataka",
    "Himanchal Pradesh": "Himachal Pradesh",
    "Maharashtra***": "Maharashtra",
    "Bihar****": "Bihar"
}

covid["State_Clean"] = covid["State_Clean"].replace(state_mapping)
print("State Names Cleaning Completed!")
print("\nUnique State Before:", covid["State/UnionTerritory"].nunique())
print("Unique States After:", covid["State_Clean"].nunique())
```

```python
# Verify Cleaned State Names

print("Cleaned State Names:")
for state in sorted(covid["State_Clean"].dropna().unique()):
    print(state)

print("\nTotal Unique Cleaned Names:",
      covid["State_Clean"].nunique())
```

```python
# Find State Names Containing Special Characters

special_states = covid[
    covid["State_Clean"].str.contains(
        r"\*|&",
        na=False,
        regex=True
    )
]["State_Clean"].unique()

print("Special State Names:")

for state in special_states:
    print(state)
```

```python
# Count Special State Records

print("Daman & Diu Records:")
print(
    covid[covid["State_Clean"] == "Daman & Diu"].shape[0]
)

print("\nMadhya Pradesh*** Records:")
print(
    covid[covid["State_Clean"] == "Madhya Pradesh***"].shape[0]
)
```

```python
# Identify Non-State Records

non_state_names = [
    "Cases being readdigned  to states",
    "Unasssigned",
    "Others"
]

non_state_data = covid[
    covid["State_Clean"].isin(non_state_names)
]
print("Non-State Records:")
display(
    non_state_data["State_Clean"].value_counts()
)

print("\nTotal Non-State Rows:", len(non_state_data))
```

```python
# Search Non-State Categories

print("Names containing 'Unassigned':")
display(
    covid[covid["State_Clean"].str.contains(
        "Unassigned", case=False, na=False
    )]["State_Clean"].unique()
)
print("Names containing 'reassigned':")
display(
    covid[covid["State_Clean"].str.contains(
        "reassigned", case=False, na=False
    )]["State_Clean"].unique()
)
```

```python
# Count Non-State Records

print("Unassigned Record:",
      covid["State_Clean"].str.contains(
          "Unassigned", case=False, na=False
      ).sum())

print("Reassigned Records:",
      covid["State_Clean"].str.contains(
          "reassigned", case=False, na=False
      ).sum())
```

```python
# Create State-wise Analysis Dataset

exclude_names = [
    "Unassigned",
    "Cases being reassigned to states"
]

covid_state = covid[
    ~covid["State_Clean"].isin(exclude_names)
].copy()

print("Original Dataset Shape:", covid.shape)
print("State-Wise Dataset Shape:", covid_state.shape)
print("Excluded Records:", len(covid) - len(covid_state))
```

```python
# Verify State-wise Dataset

print("Remaining Unassigned Records:",
      covid_state["State_Clean"].str.contains(
          "Unassigned", case=False, na=False
      ).sum())

print('Remaining Reassigned Record:',
      covid_state["State_Clean"].str.contains(
          "resassigned", case=False, na=False
      ).sum())

print("\nState-wise Dataset Shape:", covid_state.shape)
```

```python
# Count Unique States After Cleaning

total_states = covid_state["State_Clean"].nunique()
print("Total Unique State Names:", total_states)
print("\nState Names:")
for state in sorted(covid_state["State_Clean"].unique()):
    print(state)
```

```python
# State-wise Confirmed COVID Cases

state_confirmed = covid_state.groupby(
    "State_Clean"
)["Confirmed"].max().reset_index()

state_confirmed = state_confirmed.sort_values(
    by="Confirmed",
    ascending=False
)

display(state_confirmed.head(10))
```

```python
# State-wise Confirmed Cases Visualization

fig = px.bar(
    state_confirmed.head(10),
    x="Confirmed",
    y="State_Clean",
    orientation="h",
    color="Confirmed",
    text="Confirmed",
    title="Top 10 COVID-19 Affected States in India",
    labels={
        "Confirmed": "Confirmed Cases",
        "State_Clean": "State"
    }
)

fig.update_layout(
    yaxis={"categoryorder": "total ascending"},
    template="plotly_white"
)

fig.show()
```

```python
# State-wise COVID Death Analysis

state_deaths = covid_state.groupby(
    "State_Clean"
)["Deaths"].max().reset_index()

state_deaths = state_deaths.sort_values(
    by="Deaths",
    ascending=False
)
display(state_deaths.head(10))
```

```python
# Top 10 States by COVID Deaths

fig = px.bar(
    state_deaths.head(10),
    x="Deaths",
    y="State_Clean",
    orientation="h",
    color="Deaths",
    text="Deaths",
    title="Top 10 States by COVID-19 Deaths",
    template="plotly_white"
)

fig.update_layout(
    yaxis={"categoryorder": "total ascending"}
)

fig.show()
```

```python
# Top 10 States by COVID Recovery Cases

state_recovered = covid_state.groupby("State_Clean")["Cured"].max().reset_index()

state_recovered = state_recovered.sort_values(
    by="Cured",
    ascending=False
)

fig = px.bar(
    state_recovered.head(10),
    x="Cured",
    y="State_Clean",
    orientation="h",
    color="Cured",
    text="Cured",
    title="Top 10 States by COVID-19 Recovered Cases",
    template="plotly_white"
)

fig.update_layout(
    yaxis={"categoryorder": "total ascending"}
)

fig.show()
```

```python
# Calculate State-wise Recovery Rate

state_rate = covid_state.groupby("State_Clean").agg({
    "Confirmed": "max",
    "Cured": "max"
}).reset_index()

state_rate["Recovery_Rate"] = (
    state_rate["Cured"] / state_rate["Confirmed"]
) * 100

state_rate = state_rate.sort_values(
    by="Recovery_Rate",
    ascending=False
)

display(state_rate.head(10).round(2))
```

```python
# Top 10 States by Recovery Rate

fig = px.bar(
    state_rate.head(10),
    x="Recovery_Rate",
    y="State_Clean",
    orientation="h",
    color="Recovery_Rate",
    text=state_rate.head(10)["Recovery_Rate"].round(2),
    title="Top 10 States by COVID-19 Recovery Rate",
    template="plotly_white"
)

fig.update_layout(
    yaxis={"categoryorder": "total ascending"},
    xaxis_title="Recovery Rate (%)",
    yaxis_title="State / Union Territory"
)

fig.show()
```

```python
# Step 36: Vaccination Dataset Overview

print("Vaccination Dataset Shape:", vaccine.shape)

print("\nColumn Names:")
print(vaccine.columns.tolist())

print("\nDataset Information:")
vaccine.info()

print("\nFirst 5 Rows:")
display(vaccine.head())
```

```python
# Step 37: Missing Values Analysis

vaccine_missing = pd.DataFrame({
    "Missing_Values": vaccine.isnull().sum(),
    "Missing_Percentage": (
        vaccine.isnull().sum() / len(vaccine)
    ) * 100
})

vaccine_missing = vaccine_missing.sort_values(
    by="Missing_Percentage",
    ascending=False
)

display(vaccine_missing.round(2))
```

```python
# Step 38: Total Vaccination Statistics

print("Total Doses Administered:",
      vaccine["Total Doses Administered"].max())

print("First Dose Administered:",
      vaccine["First Dose Administered"].max())

print("Second Dose Administered:",
      vaccine["Second Dose Administered"].max())

print("Total Individuals Vaccinated:",
      vaccine["Total Individuals Vaccinated"].max())

print("Total Vaccination Sessions:",
      vaccine["Sessions"].max())
```

```python
# Step 39: Filter India-level Vaccination Data

india_vaccine = vaccine[
    vaccine["State"].str.strip().str.lower() == "india"
].copy()

india_vaccine = india_vaccine.sort_values(
    by="Updated On"
)

print("India Records:", india_vaccine.shape[0])

print("\nLatest India Vaccination Record:")

display(india_vaccine.tail(1)[[
    "Updated On",
    "Total Doses Administered",
    "First Dose Administered",
    "Second Dose Administered",
    "Total Individuals Vaccinated"
]])
```

```python
# Step 40: Latest Valid India Vaccination Record

india_valid = india_vaccine.dropna(
    subset=["Total Doses Administered"]
)

latest_india = india_valid.sort_values(
    by="Updated On"
).tail(1)

display(latest_india[[
    "Updated On",
    "Total Doses Administered",
    "First Dose Administered",
    "Second Dose Administered",
    "Total Individuals Vaccinated"
]])
```

```python
# Step 41: State-wise Total Vaccination

state_vaccine = vaccine[
    vaccine["State"].str.strip().str.lower() != "india"
].copy()

state_vaccine = (
    state_vaccine.groupby("State")["Total Doses Administered"]
    .max()
    .reset_index()
)

state_vaccine = state_vaccine.sort_values(
    by="Total Doses Administered",
    ascending=False
)

display(state_vaccine.head(10))
```

```python
# Step 42: Top 10 Vaccinated States

top10_vaccine = state_vaccine.head(10)

fig = px.bar(
    top10_vaccine,
    x="Total Doses Administered",
    y="State",
    orientation="h",
    color="Total Doses Administered",
    text="Total Doses Administered",
    title="Top 10 States by COVID-19 Vaccination",
    template="plotly_white"
)

fig.update_layout(
    yaxis={"categoryorder": "total ascending"},
    xaxis_title="Total Doses Administered",
    yaxis_title="State"
)

fig.show()
```

```python
# Step 43: State-wise First Dose Analysis

state_first_dose = (
    vaccine[
        vaccine["State"].str.strip().str.lower() != "india"
    ]
    .groupby("State")["First Dose Administered"]
    .max()
    .reset_index()
)

state_first_dose = state_first_dose.sort_values(
    by="First Dose Administered",
    ascending=False
)

display(state_first_dose.head(10))
```

```python
# Step 44: State-wise Second Dose Analysis

state_second_dose = (
    vaccine[
        vaccine["State"].str.strip().str.lower() != "india"
    ]
    .groupby("State")["Second Dose Administered"]
    .max()
    .reset_index()
)

state_second_dose = state_second_dose.sort_values(
    by="Second Dose Administered",
    ascending=False
)

display(state_second_dose.head(10))
```

```python
# Step 45: First Dose vs Second Dose Comparison

dose_comparison = pd.merge(
    state_first_dose,
    state_second_dose,
    on="State"
)

dose_comparison["Total_Doses"] = (
    dose_comparison["First Dose Administered"] +
    dose_comparison["Second Dose Administered"]
)

dose_comparison = dose_comparison.sort_values(
    by="Total_Doses",
    ascending=False
).head(10)

fig = px.bar(
    dose_comparison,
    x="State",
    y=[
        "First Dose Administered",
        "Second Dose Administered"
    ],
    barmode="group",
    title="First Dose vs Second Dose by State",
    template="plotly_white"
)

fig.update_layout(
    xaxis_title="State",
    yaxis_title="Number of Doses",
    xaxis_tickangle=-45
)

fig.show()
```

```python
# Step 46: Testing Dataset Overview

print("Testing Dataset Shape:", testing.shape)

print("\nColumn Names:")
print(testing.columns.tolist())

print("\nDataset Information:")
testing.info()

print("\nFirst 5 Rows:")
display(testing.head())
```

```python
# Step 47: Testing Missing Values Analysis

testing_missing = pd.DataFrame({
    "Missing_Values": testing.isnull().sum(),
    "Missing_Percentage": (
        testing.isnull().sum() / len(testing)
    ) * 100
})

testing_missing = testing_missing.sort_values(
    by="Missing_Percentage",
    ascending=False
)

display(testing_missing.round(2))
```

```python
# Step 48: COVID Testing Statistics

print("Maximum Total Samples:",
      testing["TotalSamples"].max())

print("Maximum Positive Cases:",
      testing["Positive"].max())

print("Maximum Negative Cases:",
      testing["Negative"].max())

print("\nTotal Testing Records:", len(testing))

print("Unique States:",
      testing["State"].nunique())
```

```python
# Step 49: State-wise Total Testing Analysis

state_testing = (
    testing.groupby("State")["TotalSamples"]
    .max()
    .reset_index()
)

state_testing = state_testing.sort_values(
    by="TotalSamples",
    ascending=False
)

display(state_testing.head(10))
```

```python
# Step 50: Top 10 States by COVID-19 Testing

top10_testing = state_testing.head(10)

fig = px.bar(
    top10_testing,
    x="TotalSamples",
    y="State",
    orientation="h",
    color="TotalSamples",
    text="TotalSamples",
    title="Top 10 States by COVID-19 Testing",
    template="plotly_white"
)

fig.update_layout(
    yaxis={"categoryorder": "total ascending"},
    xaxis_title="Total Samples Tested",
    yaxis_title="State"
)

fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Prepare Dashboard
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        "Top 10 States - Confirmed Cases",
        "Top 10 States - Deaths",
        "Top 10 States - Vaccination",
        "Top 10 States - Testing"
    ),
    horizontal_spacing=0.15,
    vertical_spacing=0.18
)

# 1. Confirmed Cases
df1 = state_confirmed.head(10).sort_values("Confirmed")

fig.add_trace(
    go.Bar(
        x=df1["Confirmed"],
        y=df1["State_Clean"],
        orientation="h",
        name="Confirmed Cases"
    ),
    row=1, col=1
)

# 2. Deaths
df2 = state_deaths.head(10).sort_values("Deaths")

fig.add_trace(
    go.Bar(
        x=df2["Deaths"],
        y=df2["State_Clean"],
        orientation="h",
        name="Deaths"
    ),
    row=1, col=2
)

# 3. Vaccination
df3 = state_vaccine.head(10).sort_values(
    "Total Doses Administered"
)

fig.add_trace(
    go.Bar(
        x=df3["Total Doses Administered"],
        y=df3["State"],
        orientation="h",
        name="Vaccination"
    ),
    row=2, col=1
)

# 4. Testing
df4 = state_testing.head(10).sort_values("TotalSamples")

fig.add_trace(
    go.Bar(
        x=df4["TotalSamples"],
        y=df4["State"],
        orientation="h",
        name="Testing"
    ),
    row=2, col=2
)

# Dashboard Layout
fig.update_layout(
    title_text="COVID-19 INDIA ANALYSIS DASHBOARD",
    title_x=0.5,
    height=900,
    width=1200,
    template="plotly_white",
    showlegend=False
)

fig.show()
```

```python
# Step 52: Export Interactive Dashboard

fig.write_html(
    "../covid_india_dashboard.html"
)

print("Dashboard saved successfully!")
print("File: covid_india_dashboard.html")
```
